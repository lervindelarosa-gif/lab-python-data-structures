## Exercise: Managing Customer Orders

As part of a business venture, you are starting an online store that sells various products. To ensure smooth operations, you need to develop a program that manages customer orders and inventory.

Follow the steps below to complete the exercise:

1. Define a list called `products` that contains the following items: "t-shirt", "mug", "hat", "book", "keychain".

2. Create an empty dictionary called `inventory`.

3. Ask the user to input the quantity of each product available in the inventory. Use the product names from the `products` list as keys in the `inventory` dictionary and assign the respective quantities as values.

4. Create an empty set called `customer_orders`.

5. Ask the user to input the name of three products that a customer wants to order (from those in the products list, meaning three products out of "t-shirt", "mug", "hat", "book" or "keychain". Add each product name to the `customer_orders` set.

6. Print the products in the `customer_orders` set.

7. Calculate the following order statistics:
   - Total Products Ordered: The total number of products in the `customer_orders` set.
   - Percentage of Products Ordered: The percentage of products ordered compared to the total available products.
   
   Store these statistics in a tuple called `order_status`.

8. Print the order statistics using the following format:
   ```
   Order Statistics:
   Total Products Ordered: <total_products_ordered>
   Percentage of Products Ordered: <percentage_ordered>% 
   ```

9. Update the inventory by subtracting 1 from the quantity of each product. Modify the `inventory` dictionary accordingly.

10. Print the updated inventory, displaying the quantity of each product on separate lines.

Solve the exercise by implementing the steps using the Python concepts of lists, dictionaries, sets, and basic input/output operations. 

````python
# Define a list called products
products = ["t-shirt", "mug", "hat", "book", "keychain"]
````

````python
# Create an empty dictionary
inventory = {}
````

````python
# Ask the user to input the quantity of each product available in the inventory.
# I used a for loop since its designed to automatically step through every item in a list from start to finish, until every product in the list has been processed!

for item in products:
    qty = int(input(f"Enter quantity for {item}: "))
    inventory[item] = qty
````

````python
# Create an empty set called customer_orders.
# set() function is the only way to create an empty set otherwise it would be a dictionary
customer_orders = set()
````

````python
# Ask the user to input the name of three products that a customer wants to order (from those in the products list)
# Repeat an action 3 times (asking the user for a product).
# Add each user input into your customer_orders set.
# append() is specifically for lists, while .add() is specifically for sets.
# Note I check that I could also use A while loop with a counter and seems more clean but I havent understood this complete yet

order1 = input("Enter first product: ")
customer_orders.add(order1)

order2 = input("Enter second product: ")
customer_orders.add(order2)

order3 = input("Enter third product: ")
customer_orders.add(order3)
````

````python
print("Products in customer orders:", customer_orders)
````

````python
# I need to calculate two numbers and store them in a tuple named order_status.
# What percentage of the catalog categories did the customer order?" (3/5 times 100 = 60\%$).
total_ordered = len(customer_orders)
percentage_ordered = (len(customer_orders) / len(products)) * 100
````

````python
# Store these statistics in a tuple called order_status.
# Creating a tuple in Python is as simple as putting comma-separated values inside parentheses ().
order_status = (total_ordered, percentage_ordered)
print(order_status)
````

````python
print("Order Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]}%")
````

````python
# Update the inventory by subtracting 1 from the quantity of each product. Modify the inventory dictionary accordingly.
for product in customer_orders:
    inventory[product] = inventory[product] - 1
````

````python
# Print the updated inventory, displaying the quantity of each product on separate lines..
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")
````
